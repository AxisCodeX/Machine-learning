```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
n = 500
df = pd.DataFrame({
    "Car_age": np.random.randint(1,11 ,n),
    "mileage":  np.random.randint(2,30,n) ,## in 1000 units
    
})
price = pd.Series( 5*df["mileage"] - 4*df["Car_age"]  + np.random.randint(4,70,n)) # in 10 k units)
```

```python
df["price"] = price
```

```python
X = df[["Car_age","mileage"]].to_numpy()
Y = df["price"].to_numpy()
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size = 0.2 , random_state = 42) 
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
print(model.coef_)
print(model.intercept_)
model_predict = model.predict(x_train)
r2_score = r2_score(y_train,model_predict)
print(r2_score)
```

```python
w = np.zeros(x_train.shape[1])
b = 0
dw = 0
db = 0
loss = 0
l_r = 0.001
print(x_train.shape)
print(w.shape)
print((x_train @ w).shape)
for epoch in range(40000):
    y_pred = x_train @ w  + b
    error = y_pred -  y_train
    loss = np.mean(error **2)
    dw = 2/n * (x_train.T @ error)
    db = 2/n * np.sum(error)

    w -= l_r * dw
    b -= l_r * db

print("w : ",w)
print("b:",b)
print("loss : ",loss)
```
