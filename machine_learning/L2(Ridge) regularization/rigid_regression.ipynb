```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x  = np.arange(1,10)
y = (2*x) + np.random.randint(-9,10,9)
n  = len(x)
```

```python
w = 0 
b = 0 
dw = 0
db = 0
loss = 0
lr = 0.01


for epoch in range(200):
    loss = 0
    dw = 0
    db = 0
    for xi,yi in zip(x,y):
        y_p = w* xi + b
        error = (y_p - yi)
        loss += error**2
        dw += 2/n *(xi * error)
        db += 2/n * error
    
    total_loss = 1/n * loss
    w-= lr * dw
    b-= lr * db

print("mse : ",total_loss)
print("w: ",w)
print("b : ",b)
y_pred = x*w + b
plt.title("Ordinary Linear regression")
plt.xlabel("X = Independent variable")
plt.ylabel("Y = Target")
plt.plot(x , y_pred , color="red")
plt.scatter(x,y,color="black")
plt.grid(1)
plt.show()
```
