```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.arange(1,10)
y = 5*x + np.random.randint(-50,50,9)
n = len(x)
```

```python
rs_values = np.linspace(0, 10000, 50)
rigid_loss = np.zeros(rs_values.shape,dtype=np.float32)
weights = np.zeros(rs_values.shape,dtype=np.float32)
```

```python
w = 0
b = 0
dw = 0
db = 0
lr = 0.000001
for idx,rs in enumerate(rs_values):
    w= 0
    b = 0
    for epoch in range(5000):
        dw = 0 
        db = 0
        loss = 0
        for xi,yi in zip(x,y):
            y_p = w*xi + b
            error = y_p - yi
            loss += error**2
            dw += 2/n * (xi * error)
            db += 2/n * error
        rigid_l = loss/n + rs * w**2
        dw += 2*rs*w
        w -= lr*dw
        b -= lr*db
        
    rigid_loss[idx] = rigid_l
    weights[idx] = w


plt.plot(rs_values, weights)
```
